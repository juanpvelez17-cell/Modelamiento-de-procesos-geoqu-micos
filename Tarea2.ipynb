```python
import matplotlib.pyplot as plt

# =========================
# PASO 1 — ELEGIR LOS DATOS
# Conjunto de datos sintético de ciencias de la Tierra
# =========================

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun",
         "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]

# Variable independiente: mes
# Variables dependientes: temperatura, precipitación y CO2

temperatura_c = [26.0, 26.4, 27.1, 27.8, 28.3, 28.7, 28.9, 29.0, 28.6, 28.0, 27.2, 26.5]
precipitacion_mm = [18, 22, 35, 68, 102, 145, 172, 160, 130, 95, 48, 25]
co2_ppm = [412, 414, 416, 418, 420, 421, 422, 421, 419, 417, 415, 413]

# =========================
# PASO 2 — CREAR LA GRÁFICA
# Usando matplotlib con enfoque orientado a objetos
# =========================

fig, axs = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

# Subgráfica 1: Temperatura vs tiempo
axs[0].plot(meses, temperatura_c, marker='o', linewidth=2)
axs[0].set_title("Temperatura mensual del aire")
axs[0].set_ylabel("Temperatura (°C)")
axs[0].grid(True, alpha=0.3)

# Subgráfica 2: Precipitación vs tiempo
axs[1].bar(meses, precipitacion_mm)
axs[1].set_title("Precipitación mensual")
axs[1].set_ylabel("Precipitación (mm)")
axs[1].grid(True, alpha=0.3)

# Subgráfica 3: CO₂ vs tiempo
axs[2].plot(meses, co2_ppm, marker='s', linewidth=2)
axs[2].set_title("CO₂ atmosférico mensual")
axs[2].set_xlabel("Mes")
axs[2].set_ylabel("CO₂ (ppm)")
axs[2].grid(True, alpha=0.3)

# Título general
fig.suptitle("Variación temporal de variables ambientales", fontsize=16)

# Ajuste de diseño
plt.tight_layout(rect=[0, 0, 1, 0.97])

plt.show()
```

# Resultados e interpretación — Problema 3  
## Mezcla de agua de mar y agua de acuífero calcáreo

Se modeló en PHREEQC la mezcla entre dos soluciones:

1. **Agua de acuífero calcáreo**, representada como el extremo continental o agua dulce.
2. **Agua de mar**, representada como el extremo salino.

Las mezclas se realizaron desde 0% agua de mar y 100% agua de acuífero, hasta 100% agua de mar y 0% agua de acuífero, en intervalos de 10%. El objetivo fue evaluar la evolución del índice de saturación de calcita \(SI_{Calcite}\) durante el proceso de mezcla.

## Interpretación del índice de saturación de calcita

El índice de saturación de calcita permite evaluar si el agua tiende a disolver o precipitar calcita:

\[
SI_{Calcite} < 0
\]

indica que el agua está **subsaturada** y tiende a disolver calcita.

\[
SI_{Calcite} = 0
\]

indica que el agua está en **equilibrio** con calcita.

\[
SI_{Calcite} > 0
\]

indica que el agua está **sobresaturada** y tiende a precipitar calcita.

En los resultados, las mezclas entre M0 y M50 presentan valores negativos de \(SI_{Calcite}\), por lo que están subsaturadas respecto a calcita. Esto significa que, en esas proporciones, el agua tendría tendencia a disolver calcita.

A partir de M60, el índice de saturación se vuelve positivo:

\[
SI_{Calcite} = 0.217
\]

Esto indica que desde aproximadamente 60% de agua de mar y 40% de agua de acuífero, la mezcla pasa a estar sobresaturada en calcita. Por lo tanto, en estas condiciones el sistema tiende más a precipitar calcita que a disolverla.

## Evolución del pH y la salinidad

El pH aumenta progresivamente desde el extremo de acuífero hasta el extremo marino:

\[
7.100 \rightarrow 8.180
\]

Esto muestra que el agua de mar aporta una condición más alcalina al sistema. Al mismo tiempo, la concentración de cloruro aumenta de manera casi lineal:

\[
9.594 \times 10^{-4} \rightarrow 5.040 \times 10^{-1} \, mol/kgw
\]

El cloruro puede interpretarse como un elemento conservativo en la mezcla, ya que aumenta proporcionalmente con el porcentaje de agua de mar.

## Evolución del calcio y bicarbonato

La concentración de calcio aumenta con el porcentaje de agua de mar:

\[
1.997 \times 10^{-3} \rightarrow 1.033 \times 10^{-2} \, mol/kgw
\]

Esto ocurre porque el agua de mar tiene mayor contenido de calcio que el agua de acuífero usado en la simulación.

Por el contrario, el bicarbonato disminuye progresivamente:

\[
3.533 \times 10^{-3} \rightarrow 2.115 \times 10^{-3} \, mol/kgw
\]

Esto indica que el agua de acuífero tiene una mayor concentración relativa de bicarbonato, mientras que el agua de mar aporta mayor salinidad, calcio, magnesio, sodio y cloruro.

## Respuesta a la pregunta del problema

El proceso de mezcla muestra que el agua del acuífero y las mezclas con bajo porcentaje de agua de mar están subsaturadas en calcita. Por lo tanto, estas mezclas pueden favorecer la disolución de calcita. En cambio, el agua de mar y las mezclas con mayor proporción marina se encuentran sobresaturadas en calcita, por lo que tienden a favorecer la precipitación o estabilidad de la calcita.

Específicamente:

- El agua del acuífero puro, M0, tiene \(SI_{Calcite} = -0.391\), por lo que está subsaturada en calcita.
- El agua de mar pura, M100, tiene \(SI_{Calcite} = 0.876\), por lo que está sobresaturada en calcita.
- El cambio de subsaturación a sobresaturación ocurre aproximadamente entre M50 y M60, es decir, cuando la mezcla supera cerca del 60% de agua de mar.

## Conclusión

La mezcla entre agua de acuífero calcáreo y agua de mar modifica significativamente el equilibrio geoquímico respecto a la calcita. Las mezclas con baja proporción de agua marina son subsaturadas y pueden disolver calcita, mientras que las mezclas con alta proporción de agua marina son sobresaturadas y pueden favorecer la precipitación de calcita.

El punto de transición ocurre aproximadamente entre 50% y 60% de agua de mar. Este comportamiento es relevante para procesos de karstificación litoral e intrusión salina en acuíferos costeros, ya que la mezcla de aguas con diferente composición química puede generar zonas donde la calcita se disuelve o precipita dependiendo de la proporción de mezcla.

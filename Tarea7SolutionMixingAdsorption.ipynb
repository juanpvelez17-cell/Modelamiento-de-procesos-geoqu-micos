# Problema 4 — Acuífero detrítico con intercambio catiónico

## Respuesta

La afirmación de que los moles del intercambiador \(X\) son **4 mmoles** es **falsa**.

Con los datos del problema:

\[
CEC = 75 \, meq/100g
\]

\[
\rho_b = 1.86 \, g/cm^3
\]

\[
\phi = 0.35
\]

Primero se convierte la CEC:

\[
75 \, meq/100g = 750 \, meq/kg = 0.75 \, eq/kg
\]

Luego se calcula la masa de sólido por litro de agua de poro:

\[
\frac{\rho_b}{\phi} = \frac{1.86}{0.35} = 5.31 \, kg/L
\]

Ahora se calcula la cantidad de sitios de intercambio:

\[
0.75 \, eq/kg \times 5.31 \, kg/L = 3.98 \, eq/L
\]

Como un equivalente de carga corresponde a un mol de sitios de intercambio:

\[
X \approx 3.98 \, mol/L
\]

Aproximando:

\[
X \approx 4 \, mol/L
\]

Por lo tanto, el valor correcto no es **4 mmoles**, sino aproximadamente:

\[
4 \, mol/L = 4000 \, mmol/L
\]

## Relación con PHREEQC

En PHREEQC se representa el intercambiador saturado en calcio como:

\[
CaX_2 = 2.0
\]

Esto significa:

\[
2.0 \, mol \, CaX_2 \times 2 = 4.0 \, mol \, X
\]

porque cada mol de \(CaX_2\) contiene dos sitios de intercambio \(X^-\).

## Interpretación

El agua salina que entra al acuífero interactúa con los sitios de intercambio catiónico del sedimento. Como el intercambiador está inicialmente saturado en calcio, los cationes del agua salina, especialmente \(Na^+\), \(K^+\), \(Mg^{2+}\) y \(Ca^{2+}\), pueden intercambiarse con el calcio retenido en el sólido.

Una reacción simplificada es:

\[
2Na^+ + CaX_2 \rightleftharpoons 2NaX + Ca^{2+}
\]

Este proceso modifica la composición química del agua subterránea durante la intrusión salina.

## Conclusión

La afirmación es **falsa**. Los sitios de intercambio no son 4 mmoles, sino aproximadamente:

\[
\boxed{4 \, mol/L}
\]

equivalentes a:

\[
\boxed{4000 \, mmol/L}
\]

Por eso, en PHREEQC se usa:

\[
CaX_2 = 2.0
\]

para representar 4 moles de sitios de intercambio \(X\).

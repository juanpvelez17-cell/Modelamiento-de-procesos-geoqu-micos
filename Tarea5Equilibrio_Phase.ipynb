# Se modelaron dos sistemas en PHREEQC para evaluar la disolución de calcita:

1. **Sistema cerrado al CO₂:** el agua no intercambia CO₂ con la atmósfera.
2. **Sistema abierto al CO₂:** el agua está en equilibrio con CO₂(g), simulando una condición expuesta a la atmósfera.

El objetivo fue comparar cuál de los dos sistemas permite una mayor disolución de calcita, usando como indicador principal la concentración de calcio disuelto \(Ca^{2+}\).

## Resultados obtenidos

| Sistema | pH | Ca²⁺ (mol/kgw) | HCO₃⁻ (mol/kgw) | CO₃²⁻ (mol/kgw) | C total (mol/kgw) | log PCO₂ | SI Calcite |
|---|---:|---:|---:|---:|---:|---:|---:|
| Cerrado al CO₂ | 9.9065 | 1.229 × 10⁻⁴ | 8.354 × 10⁻⁵ | 3.377 × 10⁻⁵ | 1.229 × 10⁻⁴ | -6.174 | 0.00 |
| Abierto al CO₂ | 8.2787 | 4.910 × 10⁻⁴ | 9.482 × 10⁻⁴ | 9.583 × 10⁻⁶ | 9.756 × 10⁻⁴ | -3.500 | 0.00 |

## Comparación de la disolución de calcita

La variable más importante para comparar la disolución de calcita es la concentración de calcio disuelto \(Ca^{2+}\), porque la calcita se disuelve según la reacción general:

\[
CaCO_3(s) \rightleftharpoons Ca^{2+} + CO_3^{2-}
\]

En el sistema cerrado al CO₂, la concentración de calcio fue:

\[
Ca^{2+} = 1.229 \times 10^{-4} \, mol/kgw
\]

En el sistema abierto al CO₂, la concentración de calcio fue:

\[
Ca^{2+} = 4.910 \times 10^{-4} \, mol/kgw
\]

Por lo tanto, el sistema abierto al CO₂ disolvió más calcita. La relación entre ambos valores es aproximadamente:

\[
\frac{4.910 \times 10^{-4}}{1.229 \times 10^{-4}} \approx 4
\]

Esto indica que el sistema abierto disolvió cerca de **cuatro veces más calcita** que el sistema cerrado.

## Análisis del pH

El sistema cerrado presentó un pH más alto:

\[
pH = 9.9065
\]

Mientras que el sistema abierto presentó un pH menor:

\[
pH = 8.2787
\]

Esta diferencia se debe a que, en el sistema abierto, el CO₂ atmosférico se disuelve en el agua y forma ácido carbónico:

\[
CO_2(g) \rightleftharpoons CO_2(aq)
\]

\[
CO_2(aq) + H_2O \rightleftharpoons H_2CO_3
\]

\[
H_2CO_3 \rightleftharpoons H^+ + HCO_3^-
\]

La liberación de \(H^+\) disminuye el pH y favorece la disolución de la calcita.

## Análisis del bicarbonato

El sistema abierto también presentó una concentración mucho mayor de bicarbonato:

\[
HCO_3^- = 9.482 \times 10^{-4} \, mol/kgw
\]

En comparación con el sistema cerrado:

\[
HCO_3^- = 8.354 \times 10^{-5} \, mol/kgw
\]

Este aumento confirma que el CO₂ disuelto participa activamente en el sistema carbonato, favoreciendo la formación de bicarbonato. La reacción puede representarse de forma simplificada como:

\[
CaCO_3(s) + CO_2(aq) + H_2O \rightleftharpoons Ca^{2+} + 2HCO_3^-
\]

Por esta razón, el sistema abierto al CO₂ presenta mayor concentración de \(Ca^{2+}\) y \(HCO_3^-\).

## Índice de saturación de calcita

En ambos sistemas, el índice de saturación de calcita fue aproximadamente:

\[
SI_{Calcite} = 0.00
\]

Esto indica que ambos sistemas alcanzaron equilibrio con la calcita. Sin embargo, aunque ambos estén en equilibrio, la cantidad de calcita disuelta no es la misma. El sistema abierto alcanza ese equilibrio con una concentración mayor de calcio y bicarbonato debido al aporte de CO₂.

## Conclusión

El sistema que disuelve más calcita es el **sistema abierto al CO₂**. Esto se evidencia por su mayor concentración de \(Ca^{2+}\), que fue de \(4.910 \times 10^{-4}\) mol/kgw, frente a \(1.229 \times 10^{-4}\) mol/kgw en el sistema cerrado. En términos relativos, el sistema abierto disolvió aproximadamente cuatro veces más calcita.

La mayor disolución se debe a que el CO₂ atmosférico entra al sistema acuoso, forma ácido carbónico y libera \(H^+\), lo cual favorece la disolución de la calcita y la formación de bicarbonato. Por tanto, la presencia de CO₂ aumenta la solubilidad de la calcita y modifica el equilibrio del sistema carbonato.

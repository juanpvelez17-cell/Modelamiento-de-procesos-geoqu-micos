# Resultados e interpretación — Problema 1  
## Efecto del aumento de CO₂ atmosférico sobre el pH del océano

Se modelaron tres escenarios de presión parcial de CO₂ en una solución marina simplificada usando PHREEQC:

| Escenario | log PCO₂ | pH | Carbono total C (mol/kgw) | CO₂ total (mol/kgw) |
|---|---:|---:|---:|---:|
| Condición actual | -3.50 | 8.369 | 2.400 × 10⁻³ | 2.400 × 10⁻³ |
| Escenario de emisiones moderadas | -3.25 | 8.153 | 2.499 × 10⁻³ | 2.499 × 10⁻³ |
| Escenario extremo | -3.00 | 7.925 | 2.578 × 10⁻³ | 2.578 × 10⁻³ |

## Análisis del pH

Los resultados muestran una disminución progresiva del pH a medida que aumenta la presión parcial de CO₂:

\[
8.369 \rightarrow 8.153 \rightarrow 7.925
\]

Esto indica que el aumento de CO₂ atmosférico favorece la acidificación del agua de mar. Aunque los valores de pH siguen siendo cercanos a condiciones ligeramente básicas, la disminución es significativa porque la escala de pH es logarítmica. Por lo tanto, una reducción en el pH representa un aumento en la concentración de iones \(H^+\).

## Análisis del carbono disuelto

El carbono total disuelto aumenta conforme se incrementa el CO₂:

\[
2.400 \times 10^{-3} \rightarrow 2.499 \times 10^{-3} \rightarrow 2.578 \times 10^{-3} \, mol/kgw
\]

Este comportamiento ocurre porque una mayor presión parcial de CO₂ favorece la disolución de CO₂ gaseoso en el agua. Una vez disuelto, el CO₂ participa en el sistema carbonato mediante las siguientes reacciones:

\[
CO_2(g) \rightleftharpoons CO_2(aq)
\]

\[
CO_2(aq) + H_2O \rightleftharpoons H_2CO_3
\]

\[
H_2CO_3 \rightleftharpoons H^+ + HCO_3^-
\]

La liberación de \(H^+\) explica el descenso del pH observado en los resultados.

## Interpretación ambiental

La simulación confirma que el incremento de CO₂ atmosférico produce una acidificación progresiva del agua de mar. En el escenario actual, con \(\log PCO_2 = -3.5\), el pH calculado fue de 8.369. En el escenario moderado, con \(\log PCO_2 = -3.25\), el pH disminuyó a 8.153. Finalmente, en el escenario extremo, con \(\log PCO_2 = -3.0\), el pH descendió hasta 7.925.

Este resultado es importante porque una disminución del pH modifica el equilibrio del sistema carbonato, afectando la disponibilidad de especies como \(HCO_3^-\) y \(CO_3^{2-}\). La reducción del ion carbonato puede influir en la estabilidad de minerales carbonatados como la calcita y la aragonita, fundamentales para organismos marinos calcificadores como corales, moluscos y foraminíferos.

## Conclusión

El aumento de la presión parcial de CO₂ atmosférico provoca una disminución progresiva del pH en la solución marina simulada. Además, se observa un incremento del carbono total disuelto, lo que evidencia una mayor incorporación de CO₂ al sistema acuoso. En conclusión, el modelo desarrollado en PHREEQC reproduce el comportamiento esperado de la acidificación oceánica: mayor CO₂ atmosférico genera menor pH y altera el equilibrio químico del sistema carbonato marino.

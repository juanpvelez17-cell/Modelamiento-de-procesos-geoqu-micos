```python
import pandas as pd

# Lista de meses (str) en el mismo orden de la tabla
precip_months = ["jan", "feb", "mar", "apr", "may", "jun",
                 "jul", "aug", "sept", "oct", "nov", "dec"]

# Lista de precipitación (float) en pulgadas en el mismo orden
precip_inches = [0.70, 0.75, 1.85, 2.93, 3.05, 2.02,
                 1.93, 1.62, 1.84, 1.31, 1.39, 0.84]

# Crear DataFrame
df = pd.DataFrame({
    "Mes": precip_months,
    "Precipitación (pulgadas)": precip_inches
})

# Mostrar como tabla, sin índices y bordes claros
df_style = (
    df.style
      .hide(axis="index")
      .set_table_styles([
          {"selector": "table", "props": [("border-collapse", "collapse"),
                                          ("border", "2px solid black")]},
          {"selector": "th, td", "props": [("border", "1px solid black"),
                                          ("padding", "6px"),
                                          ("text-align", "center")]}
      ])
)

df_style
```

<div style="text-align:center; font-size:16px; font-weight:700; text-transform:uppercase;">
EJERCICIO 2
</div>

```python
import pandas as pd

# Factor de conversión
inch_to_mm = 25.4

# Copia de la lista original (para no modificar precip_inches)
precip_mm = precip_inches.copy()

# Convertir cada valor en la copia
for i in range(len(precip_mm)):
    precip_mm[i] = precip_mm[i] * inch_to_mm

# Tabla comparativa (pulgadas vs mm)
df_conv = pd.DataFrame({
    "Mes": precip_months,
    "Precipitación (pulgadas)": precip_inches,
    "Precipitación (mm)": precip_mm
})

# Mostrar sin índice y con bordes claros
df_conv_style = (
    df_conv.style
      .hide(axis="index")
      .format({"Precipitación (pulgadas)": "{:.2f}", "Precipitación (mm)": "{:.2f}"})
      .set_table_styles([
          {"selector": "table", "props": [("border-collapse", "collapse"),
                                          ("border", "2px solid black")]},
          {"selector": "th, td", "props": [("border", "1px solid black"),
                                          ("padding", "6px"),
                                          ("text-align", "center")]}
      ])
)

df_conv_style
```

<div style="text-align:center; font-size:16px; font-weight:700; text-transform:uppercase;">
EJERCICIO 3
</div>

```python
# Agrupar las dos listas en una sola estructura (lista de listas)
all_precip_data = [precip_months, precip_mm]

# Verificación: imprimir y confirmar que hay 2 sublistas
print(all_precip_data)
print("Número de sublistas:", len(all_precip_data))
```

<div style="text-align:center; font-size:16px; font-weight:700; text-transform:uppercase;">
PREGUNTAS DE REFLEXIÓN
</div>

**Ventajas de agrupar datos relacionados en una sola estructura:**
Te queda todo en un solo “paquete”, más ordenado. Es más fácil llevar meses y precipitación juntos, recorrerlos al mismo tiempo, pasarlos a NumPy o a un DataFrame y evitar que se te desalineen los datos.

**Limitaciones de una lista de listas frente a una tabla:**
Una lista de listas no tiene nombres de columnas, ni tipos de datos bien controlados, ni filtros/consultas fáciles. Tampoco es tan cómoda para calcular estadísticas, ordenar por valores o hacer gráficos; en una tabla (DataFrame) eso ya viene integrado y es más claro de leer.

<div style="text-align:center; font-size:16px; font-weight:700; text-transform:uppercase;">

```python
import matplotlib.pyplot as plt

# Crear figura y ejes (ajusta el tamaño)
fig, ax = plt.subplots(figsize=(10, 5))

# Graficar datos usando all_precip_data
ax.bar(
    all_precip_data[0],      # meses
    all_precip_data[1],      # precipitación en mm
    color="blue"
)

# Configurar títulos y etiquetas
ax.set(
    title="Precipitación media mensual (mm)",
    xlabel="Mes",
    ylabel="Precipitación (mm)"
)

plt.show()
```

<div style="text-align:center; font-size:16px; font-weight:700; text-transform:uppercase;">
PREGUNTAS DE REFLEXIÓN
</div>

- **¿Por qué es importante saber cómo están organizados los datos antes de graficar?**  
  Porque si no entiendes qué es cada lista (eje X vs eje Y) puedes graficar cosas equivocadas, desordenadas o desalineadas, por ejemplo, meses que no corresponden a sus valores, por lo que esto termina en una mala interpretación por errores con los datos.

- **¿Qué limitaciones tiene trabajar con listas frente a usar un DataFrame?**  
  Con listas todo es más manual: no hay nombres de columnas, es más fácil equivocarse con índices, y es más incómodo filtrar, ordenar o calcular estadísticas. Un DataFrame ya trae estructura, etiquetas, y herramientas directas para análisis y gráficos más limpios.
